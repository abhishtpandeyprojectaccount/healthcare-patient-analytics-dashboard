## Data Analysis And Visualization

#### Which medical condition affects the most patients across the network?

```python
condition_count = df['Medical Condition']. value_counts()
print(condition_count)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(condition_count.index, condition_count.values, color='#7833FE')
plt.title('Patients by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Number of patients')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

### how is patients age distributed across the networks?

```python
df.head()
```

```python
average_age = df ['Age'].mean()
print(average_age)
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Age'], bins=20, color='Red' , edgecolor='black')
plt.title('Age Distribution by patients')
plt.xlabel('Age')
plt.ylabel('Number of patients')
plt.tight_layout()
plt.show()
```

### how has average treatment cost changed over time?

```python
average_cost= df['Billing Amount'].mean()
average_cost
```

```python
df['Admission_month']=df['Date of Admission'].dt.to_period('M')
monthly_avg_cost = df.groupby('Admission_month')['Billing Amount'].mean()
monthly_avg_cost
```

```python
plt.figure(figsize=(10, 5))
plt.plot(monthly_avg_cost.index.astype(str), monthly_avg_cost.values, color='#7833FE', marker='o', markersize=3)
plt.title('Average Treatment Cost Over Time(Monthly)')
plt.xlabel('Month')
plt.ylabel('Average Billing Amount')
plt.tight_layout()
plt.xticks(rotation=90)
plt.show()
```

### is there a relationship between patient age and treatment cost?

```python
senior_avg_cost = df.loc[df['Age']>=60,'Billing Amount'].mean()
senior_avg_cost
```

```python
sample_df= df.sample(1000, random_state=1)
plt.figure(figsize = (8, 5))
plt.scatter(sample_df['Age'], sample_df['Billing Amount'], color= 'purple',alpha= 0.5,s=15)
plt.title('Age vs Treatment Cost')
plt.xlabel('Age')
plt.ylabel('Billing Amount')
plt.tight_layout()
plt.show()
```

### which insurance provider covers the most patients?

```python
insurance_count = df['Insurance Provider']. value_counts()
insurance_count
```

```python
plt.figure(figsize=(7, 7))
plt.pie(insurance_count.values, labels=insurance_count.index, autopct='%1.1f%%', colors= ['Red', 'green', 'blue', 'purple', 'orange'])
plt.title('Patients Share by Insurance Provider')
plt.tight_layout()
plt.show()
```

### QUESTION 2---- Identify High-Cost Patients Groups

Business Scenarios: "Hospital finance team wants to identify patients who have unusually high billing amounts so they can review those cases."

```python
high_cost_patients = df.loc[df['Billing Amount']>df['Billing Amount']. quantile(0.90)]
high_cost_patients.shape[0]
```

### QUESTION 3  Flagging emergency cases with abnormal results for reviews

 Business Scenarios: "Hospital management wants to identify whose test results are abnormal,so these cases can be prioritized for review."

```python
priority_cases = df.loc[(df['Admission Type']=='Emergency')&(df['Test Results']=='Abnormal')]
priority_cases.shape[0]
```
